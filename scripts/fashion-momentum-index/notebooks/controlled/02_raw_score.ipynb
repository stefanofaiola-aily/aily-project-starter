# 2. Controlled raw FMI

Review Entities, Peers, Analyses, Financial review and Post review in Excel first. This stage calculates raw scores and pair diagnostics only. Missing components do not become neutral scores. Financial availability requires an eligible dated source/export, not a manually entered vintage.


```python
import json
import os
import subprocess
import sys
from pathlib import Path

import pandas as pd

ROOT = next(
    p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").exists()
)
OUT = Path(os.environ.get("FMI_OUTPUT_DIR", str(ROOT / "outputs/fmi-controlled")))
OUT.mkdir(parents=True, exist_ok=True)
INPUTS = OUT / "FMI_inputs.xlsx"
SNAPSHOT = OUT / "snapshot.json"


def command(*args):
    subprocess.run(
        [sys.executable, "-m", "aily_project.cli", "fmi", *map(str, args)],
        cwd=ROOT,
        check=True,
    )
```

```python
command(
    "score",
    "--inputs",
    INPUTS,
    "--snapshot",
    SNAPSHOT,
    "--output",
    OUT / "raw-score.json",
)
raw = json.loads((OUT / "raw-score.json").read_text())
display(
    pd.json_normalize(raw["results"]).filter(
        regex="brand_id|analysis_id|gaps|values.*score|values.raw_fmi|values.*growth|values.*gap_pp"
    )
)
```
