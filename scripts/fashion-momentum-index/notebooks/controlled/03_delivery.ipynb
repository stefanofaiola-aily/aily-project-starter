# 3. Display, uplift and app delivery

Refreshes the same service calculations from current reviewed inputs. Rescaling uses selected available headlines only. The scenario rate uses the rounded display score; adjusted amounts require an eligible dated baseline. Current forecasts never substitute for historical vintages. Cohort changes can change a brand's uplift; 70–98 gives all scored brands positive uplift. This is a business scenario, not predictive validation.


```python
import json
import os
import subprocess
import sys
from pathlib import Path

import pandas as pd

ROOT = next(
    p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").exists()
)
OUT = Path(os.environ.get("FMI_OUTPUT_DIR", str(ROOT / "outputs/fmi-controlled")))
OUT.mkdir(parents=True, exist_ok=True)
INPUTS = OUT / "FMI_inputs.xlsx"
SNAPSHOT = OUT / "snapshot.json"


def command(*args):
    subprocess.run(
        [sys.executable, "-m", "aily_project.cli", "fmi", *map(str, args)],
        cwd=ROOT,
        check=True,
    )
```

```python
command(
    "deliver",
    "--inputs",
    INPUTS,
    "--snapshot",
    SNAPSHOT,
    "--output",
    OUT / "weighted-delivery.xlsx",
    "--previews",
)
delivery = json.loads((OUT / "weighted-delivery.json").read_text())
display(
    pd.DataFrame(delivery["delivery"]).loc[
        :,
        [
            "brand_id",
            "name",
            "raw_fmi",
            "display_fmi",
            "fmi_status",
            "uplift_fraction",
            "adjusted_forecast",
            "forecast_status",
            "missing_reasons",
        ],
    ]
)
print(
    json.loads((OUT / "coverage.json").read_text())["prospects"],
    "prospects represented",
)
```
